```python
import pandas as pd
import numpy as np
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

import os

sns.set_theme(style="ticks")
sns.set_context("poster", font_scale=2)
sns.set_palette("Set1", 10)
sns.set_style({"font.family": "serif", "font.serif": "Times New Roman"})


if Path.cwd().name == "workdir":
    pass
elif Path.cwd().name == "proj-jailbreak-sandbox":
    os.chdir("workdir")
elif Path.cwd().name == "core":
    os.chdir("../../workdir")
else:
    print(Path.cwd())
    print("Current directory is not workdir or proj-jailbreak-sandbox. Please check the path.")
    exit(1)

print(Path.cwd())
figure_dir_path = Path(r"C:\Users\htsai\src\2024-Defend-Transferable-Jailbreak-with-Speculative-Inference\figures")
dataset_name = "RPAB"
```

```python
target_model_name_to_target_model_alias = {
    "Meta-Llama-3-70B-Instruct-AWQ": "Llama-3",
    "Qwen1.5-72B-Chat-AWQ": "Qwen-1.5",
    "Phi-3-medium-128k-instruct": "Phi-3",
}

target_model_alias_to_target_model_name = {
    v: k for k, v in target_model_name_to_target_model_alias.items()
}


def construct_dataframe(dataset_name, draft_model_name):
    result = []

    for draft_number in range(5, 35 + 1, 5):
        path = Path.cwd() / "step_12_result" / draft_model_name / f"{draft_number}" / f"{dataset_name}.pkl"
        data = pd.read_pickle(path)
        ground_truth_labels = [False] * len(data)

        for ratio in range(0, 100, 10):
            ratio_percentage = ratio / 100
            labels = list(map(lambda x: np.mean(x["labels"]) > ratio_percentage, data))
            # print(labels)
            Accuracy = accuracy_score(ground_truth_labels, labels)

            result.append(
                {
                    "Draft Number": draft_number,
                    "Ratio": ratio,
                    "Accuracy": Accuracy,
                    "Draft Model": draft_model_name,
                }
            )
    df = pd.DataFrame(result)
    return df
```

```python
def plot_Accuracy(dataframe, x, hue, palette):
    figure = sns.relplot(
        data=dataframe,
        kind="line",
        x=x,
        y="Acc (%)",
        # row="Target Model",
        col="Draft Model",
        hue=hue,
        style=hue,
        palette=palette,
        markers=True,
        markersize=26,
        linewidth=10,
        aspect=1.2,
        height=8,
        facet_kws={"margin_titles": True, "sharey": False},
    )
    # figure.set_titles("")
    # figure.set_titles(row_template="{row_name}", col_template="{col_name}")
    figure.set_titles(col_template="{col_name}")
    

    for _, ax in figure.axes_dict.items():
        ax.set_ylim(55, 105)
        ax.yaxis.set_major_locator(plt.MaxNLocator(nbins=3))
        # ax.yaxis.set_major_formatter(plt.FormatStrFormatter("%.2f"))

    sns.move_legend(figure, "lower center", bbox_to_anchor=(0.4, 1), ncol=5)

    figure.tight_layout()

    # plt.xlabel(x)
    # plt.ylabel("ASR")
```

```python
plt.clf()
for dataset_name in ["just-eval-instruct-regular"]:
    df_1 = construct_dataframe(dataset_name, "opt-125m-AWQ")
    df_2 = construct_dataframe(dataset_name, "SmolLM-135M")
    df = pd.concat([df_1, df_2])
    df["Acc (%)"] = df["Accuracy"] * 100
    plot_Accuracy(df, "Draft Number", "Ratio", "tab10")
    plt.savefig(
        figure_dir_path/ f"ASR_draft_number_{dataset_name}.pdf",
        bbox_inches="tight",
        transparent="True",
        pad_inches=0,
    )
```

```python
def construct_dataframe_for_bar(dataset_name, draft_model_name, ground_truth_guard_name, include_na: bool, PR_implementation: str):
    result = []
    for target_model_name in [
        "Meta-Llama-3-70B-Instruct-AWQ",
        "Qwen1.5-72B-Chat-AWQ",
        "Phi-3-medium-128k-instruct",
    ]:
        for generation_name in ["GCG", "AutoDAN"]:
            path = (
                Path.cwd()
                / "step_7_result"
                / dataset_name
                / target_model_name
                / generation_name
                / draft_model_name
                / f"{ground_truth_guard_name}_{0.1}.pkl"
            )
            df = pd.read_pickle(path)

            guard_name_list = (["NA"] if include_na else []) + [
                "LlamaGuardPrompt",
                PR_implementation,
                "sandbox_5",
            ]

            for guard_name in guard_name_list:
                if draft_model_name == "SmolLM-135M":
                    system_title = "Y(S)"
                elif draft_model_name == "opt-125m-AWQ":
                    system_title = "Y(O)"
                else:
                    assert False

                guard_title_dict = {
                    "NA": "N/A",
                    "LlamaGuardPrompt": "P",
                    PR_implementation: "PR",
                    "sandbox_5": system_title,
                }
                guard_title = guard_title_dict[guard_name]
                result.append(
                    {
                        "Guard": guard_title,
                        "ASR": df.loc[guard_name]["ASR"],
                        "Time": df.loc[guard_name]["average_time_for_successful_defense"],
                        "Target Model": target_model_name_to_target_model_alias[
                            target_model_name
                        ],
                        "Draft Model": draft_model_name,
                        "Generation": generation_name,
                    }
                )
    df = pd.DataFrame(result)
    return df
```

```python
def plot_bar(dataframe, x, y, palette, yscale_log=False):
    figure = sns.catplot(
        data=dataframe,
        kind="bar",
        x=x,
        y=y,
        row="Generation",
        col="Target Model",
        # hue=hue,
        # style=hue,
        palette=palette,
        aspect=1.5,
        fill=False,
        sharey=False,
        margin_titles=True,
    )
    # figure.set_titles("")
    figure.set_titles(row_template="{row_name}", col_template="{col_name}")

    for (target_model_alias, generation_name), ax in figure.axes_dict.items():
        for container in ax.containers:
            ax.set_ylim(top=ax.get_ylim()[1] * 1.05)
            ax.bar_label(container, fmt="%.1f", label_type="edge")
        # ax.yaxis.set_major_locator(plt.MaxNLocator(nbins=3))
        if yscale_log:
            ax.set_yscale("log")

        hatches = ["\\", "+", "//", ".", "|"]
        for bars, hatch in zip(ax.containers, hatches):
            for bar in bars:
                bar.set_hatch(hatch)
                
    # sns.move_legend(figure, "lower center", bbox_to_anchor=(0.4, 1), ncol=5)

    figure.tight_layout()

sns.set_context("poster", font_scale=1.3)
```

```python
plt.clf()
for ground_truth_guard_name in ["GPT", "AllTrueGuard"]:
    # for draft_model_name in ["opt-125m-AWQ", "SmolLM-135M"]:
    df_1 = construct_dataframe_for_bar(
        dataset_name, "opt-125m-AWQ", ground_truth_guard_name, False, "LlamaGuardPromptResponse"
    )
    df_2 = construct_dataframe_for_bar(
        dataset_name, "SmolLM-135M", ground_truth_guard_name, False, "LlamaGuardPromptResponse"
    )
    df = pd.concat([df_1, df_2]).drop_duplicates().reset_index(drop=True)
    df["Time (s)"] = df["Time"]
    print(df.columns)
    plot_bar(df, "Guard", "Time (s)", "deep")
    plt.savefig(
        figure_dir_path
        / f"time_bar_{ground_truth_guard_name}.pdf",
        bbox_inches="tight",
        transparent="True",
        pad_inches=0,
    )
```
